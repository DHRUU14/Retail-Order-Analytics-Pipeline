```python
import pandas as pd

df = pd.read_csv("/content/orders.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe(include="all").T
```

```python
missing = df.isnull().sum()

print("Missing values by column:")
print(missing[missing > 0])
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing_summary[missing_summary["Missing Count"] > 0]
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
print("Duplicate Order IDs:", df["Order Id"].duplicated().sum())
```

```python
# Find common placeholder values in text columns

text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    values = df[col].astype("string").str.strip().str.lower()
    mask = values.isin([
        "unknown",
        "not available",
        "n/a",
        "na",
        "none",
        "null"
    ])

    if mask.any():
        print(f"\nColumn: {col}")
        print(values[mask].value_counts())
```

```python
price_columns = ["cost price", "List Price"]

for col in price_columns:
    print(f"\n{col}")
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
print("Zero quantities:", (df["Quantity"] == 0).sum())
print("Negative quantities:", (df["Quantity"] < 0).sum())

print("\nDiscount percentage values:")
print(df["Discount Percent"].value_counts().sort_index())
```

```python
print("First 10 order dates:")
print(df["Order Date"].head(10).to_list())

print("\nDate column data type:")
print(df["Order Date"].dtype)
```

```python
zero_cost = df["cost price"] == 0
zero_list = df["List Price"] == 0

print("Both prices are zero:", (zero_cost & zero_list).sum())
print("Only cost price is zero:", (zero_cost & ~zero_list).sum())
print("Only list price is zero:", (~zero_cost & zero_list).sum())
```

```python
df.loc[
    zero_cost | zero_list,
    ["Order Id", "Order Date", "Product Id",
     "cost price", "List Price", "Quantity", "Discount Percent"]
].head(15)
```

```python
print(df["Ship Mode"].value_counts(dropna=False))
```

```python
dates = pd.to_datetime(
    df["Order Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

print("Dates that failed validation:", dates.isna().sum())
print("Earliest order date:", dates.min())
print("Latest order date:", dates.max())
```

```python
zero_price_rows = df[
    (df["cost price"] == 0) &
    (df["List Price"] == 0)
]

print("Zero-price records:", len(zero_price_rows))
print("Unique products in these records:",
      zero_price_rows["Product Id"].nunique())

zero_price_rows[
    ["Product Id", "Category", "Sub Category",
     "cost price", "List Price"]
].head(10)
```

```python
zero_product_ids = zero_price_rows["Product Id"].unique()

other_records = df[
    df["Product Id"].isin(zero_product_ids) &
    ((df["cost price"] > 0) | (df["List Price"] > 0))
]

print("Records with nonzero prices for these products:",
      len(other_records))

other_records[
    ["Product Id", "cost price", "List Price"]
].head(10)
```

```python
# Select one product that has zero prices
sample_product = zero_price_rows["Product Id"].iloc[0]

# Show all records for that product
df[df["Product Id"] == sample_product][
    ["Order Id", "Order Date", "Product Id",
     "cost price", "List Price", "Quantity",
     "Discount Percent"]
]
```

```python
df["zero_price_flag"] = (
    (df["cost price"] == 0) &
    (df["List Price"] == 0)
)

print("Flagged records:", df["zero_price_flag"].sum())
print("Original rows preserved:", len(df))
```

```python
# Create a separate working copy
df_work = df.copy()

print("Original dataset rows:", len(df))
print("Working dataset rows:", len(df_work))
print("Working dataset columns:", len(df_work.columns))
```

```python

# STEP 13: Clean and validate the working dataset

# 1. Create a clean working copy
df_clean = df_work.copy()

# 2. Preserve the original shipping values for auditing
df_clean["Ship Mode Original"] = df_clean["Ship Mode"]

# Standardize text for checking placeholders
ship_text = df_clean["Ship Mode"].astype("string").str.strip()

invalid_ship = (
    ship_text.isna()
    | ship_text.str.lower().isin(["not available", "unknown", ""])
)

# We don't know the correct shipping method, so use a proper missing value
df_clean.loc[invalid_ship, "Ship Mode"] = pd.NA

# 3. Convert order dates to a proper datetime type
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

# 4. Treat postal codes as identifiers rather than measurements
df_clean["Postal Code"] = df_clean["Postal Code"].astype("string")

# 5. Generate a data-quality report
print("========== CLEANING REPORT ==========")
print("Original rows:", len(df))
print("Cleaned working rows:", len(df_clean))
print("Original columns:", len(df.columns))
print("Cleaned columns:", len(df_clean.columns))

print("\nMissing values in cleaned shipping method:")
print(df_clean["Ship Mode"].isna().sum())

print("\nInvalid dates:", df_clean["Order Date"].isna().sum())
print("Postal Code data type:", df_clean["Postal Code"].dtype)
print("Order Date data type:", df_clean["Order Date"].dtype)

print("\nZero-price records flagged:", df_clean["zero_price_flag"].sum())
print("Exact duplicate rows:", df_clean.duplicated().sum())

print("\nCleaned dataset preview:")
display(df_clean.head())

```

```python

import pandas as pd

# STEP 14A: Build a data-quality report
quality_report = pd.DataFrame({
    "Column": df_clean.columns,
    "Data Type": [str(dtype) for dtype in df_clean.dtypes],
    "Missing Values": df_clean.isna().sum().values,
    "Unique Values": df_clean.nunique().values
})

print("========== DATA QUALITY REPORT ==========")
display(quality_report)

# STEP 14B: Validate the dataset
print("\n========== VALIDATION ==========")
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
print("Duplicate rows:", df_clean.duplicated().sum())
print("Missing shipping methods:", df_clean["Ship Mode"].isna().sum())
print("Invalid dates:", df_clean["Order Date"].isna().sum())
print("Flagged zero-price rows:", df_clean["zero_price_flag"].sum())

# STEP 14C: Save the cleaned dataset
df_clean.to_csv("/content/retail_orders_cleaned.csv", index=False)

# Save the quality report separately
quality_report.to_csv("/content/data_quality_report.csv", index=False)

print("\nFiles created:")
print("/content/retail_orders_cleaned.csv")
print("/content/data_quality_report.csv")
```
